### Training data

```python
import numpy as np 
import pandas as pd
```

```python
file = pd.read_csv("./data/documents_messy.csv")
```

```python
total_number_of_row = len(file)
total_number_of_row
```

### Handle duplicate

```python
file.duplicated()
```

```python
file = file.drop_duplicates()
```

```python
file.duplicated()
```

```python
print(file[file.duplicated()])
```

### Handle NaN

```python
print(file.isnull())
```

```python
print(file.isnull().sum())
```

```python
file = file.dropna()
file
```

```python
file.isnull()
```

```python
print(file['title'].dtype)
```

```python
file["category"].str.strip().str.lower().replace({"n/a": pd.NA, "nan": pd.NA, "none": pd.NA, "": pd.NA})
```

```python
file = file.dropna(subset=["category"])
```

### Column

```python
file.columns
```

```python
file = file.drop(columns=['created_at'])
```

```python
left_raws = len(file)
left_raws
```

```python
accepted_count = left_raws
accepted_count
```

```python
rejected_count = total_number_of_row - left_raws
rejected_count
```

### Finalizing CSV

```python
file['status'] = 'complete'
```

```python
file['accepted_count'] = accepted_count
```

```python
file['rejected_count'] = rejected_count
```
