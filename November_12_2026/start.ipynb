### NumPy

```python
import numpy as np
```

```python
w = np.arange(1,11,2)
h = np.linspace(1,10,5)
```

```python
print(w)
print(h)
```

```python
def solve(w: list,h:list) -> float:
    result = 0
    for i, j in zip(w, h):
        result += i * j
    
    return result
```

```python
ans = solve(w, h)
ans.astype(int)
```
