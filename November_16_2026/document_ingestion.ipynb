### Document ingestion

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_community.document_loaders import TextLoader
```

```python
from pathlib import Path
```

```python
def load_pdf(file_path: str = "./data/a2.pdf"):

    path = file_path.strip()

    # Check file existence
    if not Path(path).is_file():
        print("File path not found")
        return None

    # Check extension
    if Path(path).suffix.lower() != ".pdf":
        print("This file is not a PDF")
        return None

    # Initialize loader
    loader = PyPDFLoader(
        file_path=path,
        headers=None,
        password=None
    )

    docs = []

    # Extract PDF content
    for doc in loader.lazy_load():

        doc.page_content = doc.page_content.strip()

        if not doc.page_content:
            continue

        docs.append(doc)

        print("Extracted text:")
        print(doc.page_content[:100])

        print("Metadata:")
        print(doc.metadata)

    if not docs:
        print("No extractable text found")
        return None

    return docs

load_pdf()
```

### Chunking

#### 1. Fixed size chunking

```python
from langchain_text_splitters import CharacterTextSplitter

text = "LangChain simplifies AI workflows. It enables advanced retrieval-augmented generation systems for NLP tasks."
splitter = CharacterTextSplitter(chunk_size=50, chunk_overlap=10, separator=" ")
chunks = splitter.split_text(text)
print("chunks:", len(chunks))
print(chunks)
```

#### 2. Recursive Character chunking

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text = """LangChain supports modular pipelines for AI workflows.  
These workflows include document loading, chunking, retrieval, and LLM integration.  
LangChain simplifies AI model deployment."""
splitter = RecursiveCharacterTextSplitter(chunk_size=50, chunk_overlap=10)
chunks = splitter.split_text(text)
print(chunks)
```

#### 3. Sentence/Paragraph chunking 

```python
text = """LangChain simplifies AI processing.  
It uses embeddings for efficient retrieval.  
This enables smarter question-answering systems."""
sentences = text.split("\n")
print(sentences)
```

#### 4. Python Code chunking

```python
from langchain_text_splitters import PythonCodeTextSplitter

text = """def add(a, b):\n    return a + b\n\ndef subtract(a, b):\n    return a - b"""
splitter = PythonCodeTextSplitter(chunk_size=200)
chunks = splitter.split_text(text)
print(chunks)
```

#### 5. Agentic Chunking

```python
from langchain_text_splitters import CharacterTextSplitter

text = """CharacterTextSplitter does not use an LLM or understand semantic meaning. It splits text using a separator (by default, \n\n) and combines the resulting pieces into chunks.

If you want splitting based on semantic meaning, LangChain provides SemanticChunker, which uses embeddings to determine semantic similarity.

My recommendation: First understand CharacterTextSplitter, RecursiveCharacterTextSplitter, and then SemanticChunker. They represent three different approaches to document chunking."""

splitter = CharacterTextSplitter(
    chunk_size=50,
    chunk_overlap=10,
    separator=" "
)
# LLM agent dynamically groups by meaning
chunks = splitter.split_text(text)
print(chunks)
```

#### 6. LLM-based semantic chunking

```python
from langchain_groq import ChatGroq
import os
from dotenv import load_dotenv

load_dotenv()
```

```python
llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0.3,
    groq_api_key=os.getenv("GROQ_API_KEY")
)
```

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
from langchain_core.prompts import ChatPromptTemplate

text = "AI revolutionizes industries. AI applications range from healthcare to finance."

prompt = ChatPromptTemplate.from_messages([
    ("system", """
    You are a professional semantic chunking model.

    Your task is to divide the provided text into smaller chunks
    based on meaning and context.

    Follow these rules:
    1. Keep related sentences together.
    2. Do not break sentences unnecessarily.
    3. Each chunk should contain a complete idea.
    4. Maintain the original meaning.
    5. Do not add or remove information.
    6. Return the chunks as a numbered list.
    """),
    ("user", "Divide the following text into meaningful chunks:\n\n{text}")
])
```

```python
chain = prompt | llm

response = chain.invoke({
    "text": text
})

print(response.content)
```

```python
import hashlib

def generate_content_hash(content: str) -> str:
    return hashlib.sha256(
        content.encode("utf-8")
    ).hexdigest()
```

### chunk id  creates no duplicate chunks

```python
def generate_hash(text):
    return hashlib.sha256(
        text.encode("utf-8")
    ).hexdigest()


chunks_1 = [
    "AI is transforming the world.",
    "Machine learning is a subset of AI.",
    "Python is used for AI development."
]

chunks_2 = [
    "AI is transforming the world.",
    "Deep learning uses neural networks.",
    "Python is used for AI development."
]

hash_map = {}

for i, chunk in enumerate(chunks_1):
    chunk_hash = generate_hash(chunk)
    hash_map[chunk_hash] = i

for j, chunk in enumerate(chunks_2):

    chunk_hash = generate_hash(chunk)

    if chunk_hash in hash_map:
        print(
            f"Chunk {j} matches "
            f"Document 1 Chunk {hash_map[chunk_hash]}"
        )
    else:
        print(f"Chunk {j} is new")
```

### Document ingestion + chunking

```python
hash_map = {}
unique_chunks = []
```

```python
def chunking(text):
    splitter = RecursiveCharacterTextSplitter(
        chunk_size=500,
        chunk_overlap=100
    )

    return splitter.split_text(text)
```

```python
def generate_hash(text):
    return hashlib.sha256(
        text.encode("utf-8")
    ).hexdigest()
```

```python
def isDuplicate(text):
    content_hash = generate_hash(text)

    if content_hash in hash_map:
        return True

    hash_map[content_hash] = len(hash_map)
    return False
```

```python

def merging(docs):
    results = []

    for doc in docs:

        chunks = chunking(doc.page_content)

        for chunk in chunks:

            if isDuplicate(chunk):
                print("Duplicate chunk found")
                continue

            results.append(chunk)
            unique_chunks.append(chunk)

    return results
```

```python

# For PDF files
def load_pdf(path: str):

    path = Path(path)

    if not path.is_file():
        print("File path not found")
        return None

    loader = PyPDFLoader(str(path))

    docs = []

    for doc in loader.lazy_load():

        doc.page_content = doc.page_content.strip()

        if not doc.page_content:
            continue

        docs.append(doc)

    if not docs:
        print("No extractable text found")
        return None

    return merging(docs)
```

```python
def load_txt(path: str):

    path = Path(path)

    if not path.is_file():
        print("File path not found")
        return None

    loader = TextLoader(
        str(path),
        encoding="utf-8"
    )

    documents = loader.load()

    return merging(documents)
```

```python
# Load all files
def load_all_files(folder_path):

    folder_path = Path(folder_path)

    if not folder_path.is_dir():
        print("Folder path not found")
        return

    for file_path in folder_path.iterdir():

        if not file_path.is_file():
            continue

        extension = file_path.suffix.lower()

        if extension == ".txt":
            temp = load_txt(file_path)

        elif extension == ".pdf":
            temp = load_pdf(file_path)

        else:
            print(f"Unsupported file extension: {file_path.name}")
            continue

        print(f"\nFile: {file_path.name}")
        print(f"Unique chunks: {len(temp) if temp else 0}")
```

```python
load_all_files("./data")
print("\nTotal unique chunks:", len(unique_chunks))
print("Total hashes:", len(hash_map))
```
